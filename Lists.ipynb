```python
lst=list(map(int,input().split()))
present=0
absent=0
i=0
while i<len(lst):
    if lst[i]==0:
        absent+=1
        i+=1
    elif lst[i]==1:
        present+=1
        i+=1
    else:
        print("invalid")
print(present)
print("present percentage",present/len(lst)*100)
        
        
```
