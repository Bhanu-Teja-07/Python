```python
#types of attributes
#1.class attributes

class Student:
    school_name="XYZ"
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def details(self):
        return f"my name is {self.name} and my age is {self.age}"

s1=Student("Ravi",25)
s2=Student("Ramu",26)
print(s1.name)
print(s2.age)
print(s1.details())
print(s1.school_name)
print(s2.school_name)
```

```python
class cal:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def add(self):
        return self.x+self.y
    def sub(self):
        return self.x-self.y
    def mul(self):
        return self.x*self.y
    def div(self):
        try:
            return self.x//self.y
        except Exception as e:
            return f"error: {e}"
c1=cal(10,20)
c2=cal(30,50)
c3=cal(10,0)
print(c1.add())
print(c2.add())
print(c3.add(),c3.sub())
print(c3.div())
```
