```python
from datetime import datetime,time,date,timedelta
currtime=datetime.now()
print(currtime)
curtime=currtime.time()
print(curtime)
current=currtime.date()
print(current)
```

```python
print(currtime.day)
print(currtime.month)
print(currtime.year)
```

```python
print(currtime.hour)
print(currtime.minute)
print(currtime.second)
print(currtime.microsecond)
```

```python
print(datetime.max)
print(current.max)
print(datetime.min)
print(current.min)
```

```python
print(currtime.strftime("%d-%m-%Y"))
print(currtime.strftime("%A"))
print(currtime.strftime("%I:%M"))
```

```python
future=current+timedelta(weeks=52)
print(future)
```
