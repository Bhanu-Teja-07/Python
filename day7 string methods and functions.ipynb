#string opertions
1.string concatenation
2.string repeatation

```python
s1="code"
s2="gnan"
print(s1+s2)
print((s1+" ")*5)
print(((s1+" ")*5)+((s2+" ")*5))
```

```python
s="Super"
print(s.upper())
print(s.lower())
print(s.title())
print(s.capitalize())
print(s.swapcase())
```

```python
s="    Integer number   "
print(s.replace("Integer", "float"))
print(s.split("e"))
print(s.strip())
print(s.lstrip())
print(s.rstrip())
```

```python
s="Prince123"
print(s.isupper())
print(s.islower())
print(s.isdigit())
print(s.isalpha())
print(s.isalnum())
print(s.istitle())
print(" ".join(s))
```
