```python
def even_or_odd(x):
    if x%2==0:
        return True
    else:
        return False

x=int(input())
result=even_or_odd(10)
if result==True:
    print("even")
else:
    print("odd")
```

#varible length of arguments

```python
def varsum(x,*y):
    print(type(y))
    print(y)
    s=0
    for i in y:
        s+=i
    return s+x
print(varsum(10,20,30,40,50))
print(varsum(10,20,30,40,50,1,2,3,4,5,6))
```
