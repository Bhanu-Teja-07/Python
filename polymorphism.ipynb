```python
#method overriding
class re:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show(self):
        return f"my name is {self.name} and i am {self.age} years old"
class emp(re):
    def __init__(self,name,age):
        re.__init__(name,age)
    def show(self):
        return f"my name is {self.name} and i am {self.age} years old"
e=re("ram",25)
p=emp("ravi",20)
print(p.show())
```

```python
#method overloading
class math:
    def add(self,a,b,c=0,d=0):
        return a+b+c+d
m=math()
print(m.add(10,20))
print(m.add(10,20,30))
print(m.add(10,20,30,40))
```
