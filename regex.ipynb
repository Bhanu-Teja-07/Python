```python
import re
s=input()
x=re.match("[a-zA-Z]+",s)
print(x)
print(x.group())
print(x.span())
print(re.findall("the",s))
```

```python
import re
s=input()
x=re.search("t",s)
print(x)
print(x.group())
print(x.span())
```

```python
import re
s=input()
x=re.fullmatch(r"^[A-Z]{5}[0-9]{4}[A-Z]$",s)
if x:
    print("pan number valid")
else:
    print("pan number invalid")
```

```python
s=input()
x=re.fullmatch(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9]+\.[a-z]{2,3}$",s)
if x:
    print("email valid")
else:
    print("email invalid")
```

```python
s=input()
dat=re.fullmatch(r"(0[1-9]|[12][0-9]|3[0-1])+\-(0[1-9]|1[0-2])+\-\d{4}$",s)
if dat:
    print("date valid")
else:
    print("date invalid")
```
