```python
class employee:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def print_details(self):
        return f"the name of the person {self.name} and age is {self.age}"
class company(employee):
    def __init__(self,name,age,comp):
        super().__init__(name,age)
        self.comp=comp
    def show(self):
        return f"the name of the person is {self.name} and the age is {self.age} and works in {self.comp}"
s=company("raju",25,"TCS")
print(s.print_details())
```
