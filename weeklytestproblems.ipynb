```python
m=int(input())
if m>90:
    print("Grade A")
elif m<90 and m>=75:
    print("Grade B")
elif m<75 and m>=60:
    print("Grade C")
elif m<60 and m>=50:
    print("Grade D")
else:
    print("Fail")
```

```python
n=int(input())
rev=0
rem=0
while n>0:
    rem=n%10
    rev=rev*10+rem
    n//=10
print(rev)
```
